```python
# ============================================================
# CELL 9 — LOAD FINBERT
# ============================================================

import torch
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification
)

MODEL_NAME = "ProsusAI/finbert"

DEVICE = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Device:", DEVICE)

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))


tokenizer = AutoTokenizer.from_pretrained(
    MODEL_NAME
)

model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME,
    num_labels=3,
    ignore_mismatched_sizes=True
)

model.to(DEVICE)

print("\n✓ FinBERT loaded")
print("Parameters:", sum(p.numel() for p in model.parameters()))
```

```python
# ============================================================
# CELL 11 — TRAINING CONFIGURATION
# ============================================================

SEED = 42

MAX_LENGTH = 128
BATCH_SIZE = 16
LEARNING_RATE = 2e-5
EPOCHS = 3
WEIGHT_DECAY = 0.01
WARMUP_RATIO = 0.1

MODEL_DIR = (
    PROJECT_ROOT
    / "models"
)

MODEL_DIR.mkdir(
    parents=True,
    exist_ok=True
)

print("Training configuration")
print("-" * 40)
print("Model:", MODEL_NAME)
print("Max length:", MAX_LENGTH)
print("Batch size:", BATCH_SIZE)
print("Learning rate:", LEARNING_RATE)
print("Epochs:", EPOCHS)
print("Weight decay:", WEIGHT_DECAY)
print("Device:", DEVICE)
```

```python
# ============================================================
# CELL 12 — TOKENIZATION PIPELINE
# ============================================================

from torch.utils.data import Dataset, DataLoader


class FinancialDataset(Dataset):

    def __init__(self, dataframe, tokenizer, max_length=128):
        self.texts = dataframe["text"].astype(str).tolist()
        self.labels = dataframe["label"].astype(int).tolist()
        self.tokenizer = tokenizer
        self.max_length = max_length

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, idx):

        encoding = self.tokenizer(
            self.texts[idx],
            truncation=True,
            padding="max_length",
            max_length=self.max_length,
            return_tensors="pt"
        )

        return {
            "input_ids": encoding["input_ids"].squeeze(0),
            "attention_mask": encoding["attention_mask"].squeeze(0),
            "labels": torch.tensor(
                self.labels[idx],
                dtype=torch.long
            )
        }


# ------------------------------------------------------------
# Create FiQA datasets
# ------------------------------------------------------------

fiqa_train_dataset = FinancialDataset(
    fiqa_train,
    tokenizer,
    MAX_LENGTH
)

fiqa_val_dataset = FinancialDataset(
    fiqa_val,
    tokenizer,
    MAX_LENGTH
)

fiqa_test_dataset = FinancialDataset(
    fiqa_test,
    tokenizer,
    MAX_LENGTH
)

print("FiQA datasets created")

print("Train:", len(fiqa_train_dataset))
print("Validation:", len(fiqa_val_dataset))
print("Test:", len(fiqa_test_dataset))
```

```python
# ============================================================
# CELL 13 — DATALOADERS
# ============================================================

fiqa_train_loader = DataLoader(
    fiqa_train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=2,
    pin_memory=True
)

fiqa_val_loader = DataLoader(
    fiqa_val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

fiqa_test_loader = DataLoader(
    fiqa_test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

print("Train batches:", len(fiqa_train_loader))
print("Validation batches:", len(fiqa_val_loader))
print("Test batches:", len(fiqa_test_loader))
```

```python
# ============================================================
# CELL 15 — TRAINING UTILITIES
# ============================================================

import copy
import time
import numpy as np

from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support,
    classification_report,
    confusion_matrix
)

from torch.optim import AdamW
from tqdm.auto import tqdm


def evaluate_model(model, dataloader, device):
    """
    Evaluate model on validation/test data.
    """

    model.eval()

    all_predictions = []
    all_labels = []

    total_loss = 0.0
    total_samples = 0

    criterion = torch.nn.CrossEntropyLoss()

    with torch.no_grad():

        for batch in tqdm(
            dataloader,
            desc="Evaluating",
            leave=False
        ):

            input_ids = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)
            labels = batch["labels"].to(device)

            outputs = model(
                input_ids=input_ids,
                attention_mask=attention_mask,
                labels=labels
            )

            loss = outputs.loss

            predictions = torch.argmax(
                outputs.logits,
                dim=1
            )

            batch_size = labels.size(0)

            total_loss += (
                loss.item() * batch_size
            )

            total_samples += batch_size

            all_predictions.extend(
                predictions.cpu().numpy()
            )

            all_labels.extend(
                labels.cpu().numpy()
            )

    avg_loss = total_loss / total_samples

    accuracy = accuracy_score(
        all_labels,
        all_predictions
    )

    precision, recall, f1, _ = (
        precision_recall_fscore_support(
            all_labels,
            all_predictions,
            average="macro",
            zero_division=0
        )
    )

    return {
        "loss": avg_loss,
        "accuracy": accuracy,
        "precision": precision,
        "recall": recall,
        "f1": f1,
        "predictions": np.array(all_predictions),
        "labels": np.array(all_labels)
    }


def train_model(
    model,
    train_loader,
    val_loader,
    device,
    epochs=3,
    learning_rate=2e-5,
    weight_decay=0.01,
    save_path=None,
    patience=2
):

    optimizer = AdamW(
        model.parameters(),
        lr=learning_rate,
        weight_decay=weight_decay
    )

    best_f1 = -1.0
    best_state = None
    patience_counter = 0

    history = []

    print("=" * 60)
    print("TRAINING")
    print("=" * 60)

    for epoch in range(epochs):

        print(
            f"\nEpoch {epoch + 1}/{epochs}"
        )

        start_time = time.time()

        # ----------------------------------------------------
        # Training
        # ----------------------------------------------------

        model.train()

        total_train_loss = 0.0
        total_train_samples = 0

        progress_bar = tqdm(
            train_loader,
            desc="Training"
        )

        for batch in progress_bar:

            input_ids = batch["input_ids"].to(
                device,
                non_blocking=True
            )

            attention_mask = batch[
                "attention_mask"
            ].to(
                device,
                non_blocking=True
            )

            labels = batch["labels"].to(
                device,
                non_blocking=True
            )

            optimizer.zero_grad(
                set_to_none=True
            )

            outputs = model(
                input_ids=input_ids,
                attention_mask=attention_mask,
                labels=labels
            )

            loss = outputs.loss

            loss.backward()

            optimizer.step()

            batch_size = labels.size(0)

            total_train_loss += (
                loss.item() * batch_size
            )

            total_train_samples += batch_size

            progress_bar.set_postfix(
                loss=f"{loss.item():.4f}"
            )

        train_loss = (
            total_train_loss
            / total_train_samples
        )

        # ----------------------------------------------------
        # Validation
        # ----------------------------------------------------

        val_metrics = evaluate_model(
            model,
            val_loader,
            device
        )

        epoch_time = (
            time.time() - start_time
        )

        print(
            f"\nTrain Loss : {train_loss:.4f}"
        )

        print(
            f"Val Loss   : {val_metrics['loss']:.4f}"
        )

        print(
            f"Val Acc    : {val_metrics['accuracy']:.4f}"
        )

        print(
            f"Val F1     : {val_metrics['f1']:.4f}"
        )

        print(
            f"Time       : {epoch_time:.1f}s"
        )

        history.append({
            "epoch": epoch + 1,
            "train_loss": train_loss,
            "val_loss": val_metrics["loss"],
            "val_accuracy": val_metrics["accuracy"],
            "val_precision": val_metrics["precision"],
            "val_recall": val_metrics["recall"],
            "val_f1": val_metrics["f1"],
            "time_seconds": epoch_time
        })

        # ----------------------------------------------------
        # Best model checkpoint
        # ----------------------------------------------------

        if val_metrics["f1"] > best_f1:

            best_f1 = val_metrics["f1"]

            best_state = copy.deepcopy(
                model.state_dict()
            )

            patience_counter = 0

            print(
                f"✓ New best model "
                f"(F1 = {best_f1:.4f})"
            )

            if save_path is not None:

                torch.save(
                    {
                        "model_state_dict": best_state,
                        "best_val_f1": best_f1,
                        "epoch": epoch + 1
                    },
                    save_path
                )

                print(
                    f"✓ Checkpoint saved → "
                    f"{save_path}"
                )

        else:

            patience_counter += 1

            print(
                f"No improvement "
                f"({patience_counter}/{patience})"
            )

            if patience_counter >= patience:

                print(
                    "\nEarly stopping."
                )

                break

    # --------------------------------------------------------
    # Restore best model
    # --------------------------------------------------------

    if best_state is not None:

        model.load_state_dict(
            best_state
        )

    print("\n" + "=" * 60)
    print(
        f"BEST VALIDATION F1: {best_f1:.4f}"
    )
    print("=" * 60)

    return model, pd.DataFrame(history)
```

```python
# ============================================================
# CELL 16 — TRAIN FiQA EXPERT
# ============================================================

fiqa_model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME,
    num_labels=3,
    ignore_mismatched_sizes=True
)

fiqa_model.to(DEVICE)

fiqa_checkpoint = (
    MODEL_DIR / "fiqa_expert.pt"
)

fiqa_model, fiqa_history = train_model(
    model=fiqa_model,
    train_loader=fiqa_train_loader,
    val_loader=fiqa_val_loader,
    device=DEVICE,
    epochs=EPOCHS,
    learning_rate=LEARNING_RATE,
    weight_decay=WEIGHT_DECAY,
    save_path=fiqa_checkpoint,
    patience=2
)

display(fiqa_history)
```

```python
# ============================================================
# CELL 18 — GENERIC EXPERT TRAINING PIPELINE
# ============================================================

def prepare_expert_loaders(
    train_df,
    val_df,
    test_df=None,
    tokenizer=tokenizer,
    max_length=128,
    batch_size=16
):
    """
    Create PyTorch DataLoaders for an expert.
    """

    train_dataset = FinancialDataset(
        train_df,
        tokenizer,
        max_length
    )

    val_dataset = FinancialDataset(
        val_df,
        tokenizer,
        max_length
    )

    train_loader = DataLoader(
        train_dataset,
        batch_size=batch_size,
        shuffle=True,
        num_workers=2,
        pin_memory=True
    )

    val_loader = DataLoader(
        val_dataset,
        batch_size=batch_size,
        shuffle=False,
        num_workers=2,
        pin_memory=True
    )

    test_loader = None

    if test_df is not None:

        test_dataset = FinancialDataset(
            test_df,
            tokenizer,
            max_length
        )

        test_loader = DataLoader(
            test_dataset,
            batch_size=batch_size,
            shuffle=False,
            num_workers=2,
            pin_memory=True
        )

    return (
        train_loader,
        val_loader,
        test_loader
    )
```
