```python
# ============================================================
# CELL 14 — NORMALIZE FiQA
# ============================================================

print("=" * 60)
print("NORMALIZING FiQA")
print("=" * 60)

# ------------------------------------------------------------
# FiQA schema
# ------------------------------------------------------------
# _id
# sentence
# target
# aspect
# score
# type
#
# We use:
# sentence -> text
# score    -> sentiment
# ------------------------------------------------------------


# Create normalized dataframe
fiqa_normalized = fiqa_train[
    ["sentence", "score"]
].copy()


# Rename columns
fiqa_normalized.columns = [
    "text",
    "score"
]


# Add source
fiqa_normalized["source"] = "fiqa"


# ------------------------------------------------------------
# Clean text
# ------------------------------------------------------------

fiqa_normalized["text"] = (
    fiqa_normalized["text"]
    .astype(str)
    .str.strip()
)


# Remove empty text
fiqa_normalized = fiqa_normalized[
    (fiqa_normalized["text"] != "") &
    (fiqa_normalized["text"] != "nan")
].copy()


# ------------------------------------------------------------
# Convert FiQA continuous score
# to common 3-class sentiment
#
# score < -0.1       -> negative = 0
# -0.1 <= score <= .1 -> neutral = 1
# score > 0.1        -> positive = 2
# ------------------------------------------------------------

def fiqa_score_to_label(score):

    if score < -0.1:
        return 0

    elif score <= 0.1:
        return 1

    else:
        return 2


fiqa_normalized["label"] = (
    fiqa_normalized["score"]
    .apply(fiqa_score_to_label)
)


# ------------------------------------------------------------
# Remove original score from final unified schema
# ------------------------------------------------------------

fiqa_normalized = fiqa_normalized[
    ["text", "label", "source"]
].reset_index(drop=True)


# ------------------------------------------------------------
# Verification
# ------------------------------------------------------------

print("\nShape:")
print(fiqa_normalized.shape)

print("\nColumns:")
print(fiqa_normalized.columns.tolist())

print("\nLabel distribution:")
print(
    fiqa_normalized["label"]
    .value_counts()
    .sort_index()
)

print("\nSource distribution:")
print(
    fiqa_normalized["source"]
    .value_counts()
)

print("\nSample:")
display(
    fiqa_normalized.head(10)
)
```

```python
# ============================================================
# CELL 15 — NORMALIZE ALL FiQA SPLITS
# ============================================================

def normalize_fiqa(df, split_name):

    result = df[
        ["sentence", "score"]
    ].copy()

    result.columns = [
        "text",
        "score"
    ]

    result["source"] = "fiqa"

    # Clean text
    result["text"] = (
        result["text"]
        .astype(str)
        .str.strip()
    )

    # Remove empty text
    result = result[
        (result["text"] != "") &
        (result["text"] != "nan")
    ].copy()

    # Convert continuous FiQA score → 3 classes
    result["label"] = result["score"].apply(
        fiqa_score_to_label
    )

    # Final schema
    result = result[
        ["text", "label", "source"]
    ].reset_index(drop=True)

    print(
        f"{split_name:10} → {result.shape}"
    )

    return result


# Normalize each split separately

fiqa_train_normalized = normalize_fiqa(
    fiqa_train,
    "Train"
)

fiqa_valid_normalized = normalize_fiqa(
    fiqa_valid,
    "Validation"
)

fiqa_test_normalized = normalize_fiqa(
    fiqa_test,
    "Test"
)
```

```python
# ============================================================
# CELL 18 — NORMALIZE TWITTER
# ============================================================

TWITTER_TEXT_COLUMN = "text"
TWITTER_LABEL_COLUMN = "label"


def normalize_twitter(df, split_name):

    result = df[
        [
            TWITTER_TEXT_COLUMN,
            TWITTER_LABEL_COLUMN
        ]
    ].copy()

    result.columns = [
        "text",
        "label"
    ]

    result["source"] = "twitter"

    # Clean text
    result["text"] = (
        result["text"]
        .astype(str)
        .str.strip()
    )

    # Remove empty text
    result = result[
        (result["text"] != "") &
        (result["text"] != "nan")
    ].copy()

    # --------------------------------------------------------
    # Handle labels
    # --------------------------------------------------------

    # If labels are already numeric 0/1/2
    if pd.api.types.is_numeric_dtype(
        result["label"]
    ):

        result["label"] = (
            result["label"]
            .astype(int)
        )

    # If labels are strings
    else:

        result["label"] = (
            result["label"]
            .astype(str)
            .str.lower()
            .str.strip()
            .map({
                "negative": 0,
                "neutral": 1,
                "positive": 2
            })
        )

    # Check for invalid labels
    invalid = result[
        ~result["label"].isin([0, 1, 2])
    ]

    if len(invalid) > 0:

        print(
            f"⚠ {split_name}: "
            f"{len(invalid)} invalid labels found"
        )

        print(
            invalid["label"]
            .value_counts(dropna=False)
        )

        raise ValueError(
            "Twitter label mapping needs to be checked."
        )

    result["label"] = (
        result["label"]
        .astype(int)
    )

    return result[
        ["text", "label", "source"]
    ].reset_index(drop=True)


twitter_train_normalized = normalize_twitter(
    twitter_train,
    "Train"
)

twitter_valid_normalized = normalize_twitter(
    twitter_valid,
    "Validation"
)


print(
    "Twitter Train:",
    twitter_train_normalized.shape
)

print(
    "Twitter Validation:",
    twitter_valid_normalized.shape
)
```

```python
# ============================================================
# CELL 21 — NORMALIZE FINANCIALPHRASEBANK
# ============================================================

phrasebank_records = []

for line in phrasebank_lines:

    line = line.strip()

    if not line:
        continue

    # Sentence and sentiment are separated by @
    parts = line.rsplit("@", 1)

    if len(parts) != 2:
        continue

    text, sentiment = parts

    text = text.strip()
    sentiment = sentiment.strip().lower()

    phrasebank_records.append({
        "text": text,
        "sentiment": sentiment
    })


phrasebank_normalized = pd.DataFrame(
    phrasebank_records
)


print("Parsed rows:", phrasebank_normalized.shape)

print("\nOriginal labels:")
print(
    phrasebank_normalized["sentiment"]
    .value_counts()
)


# ------------------------------------------------------------
# Map sentiment → common HACE labels
# ------------------------------------------------------------

phrasebank_normalized["label"] = (
    phrasebank_normalized["sentiment"]
    .map({
        "negative": 0,
        "neutral": 1,
        "positive": 2
    })
)


# Check invalid labels
invalid = phrasebank_normalized[
    phrasebank_normalized["label"].isna()
]

print("\nInvalid labels:", len(invalid))

if len(invalid) > 0:
    print(invalid["sentiment"].value_counts())


# Remove sentiment text column
phrasebank_normalized = (
    phrasebank_normalized[
        ["text", "label"]
    ]
    .copy()
)

phrasebank_normalized["source"] = "phrasebank"

phrasebank_normalized["label"] = (
    phrasebank_normalized["label"]
    .astype(int)
)

phrasebank_normalized = (
    phrasebank_normalized
    .reset_index(drop=True)
)


print("\nFinal shape:")
print(phrasebank_normalized.shape)

display(
    phrasebank_normalized.head()
)
```

```python
# ============================================================
# CELL 30 — DUPLICATE / LEAKAGE AUDIT
# ============================================================

print("=" * 60)
print("DUPLICATE / LEAKAGE AUDIT")
print("=" * 60)


def normalize_for_duplicate_check(text):
    return (
        str(text)
        .lower()
        .strip()
    )


# ------------------------------------------------------------
# Create datasets to audit
# ------------------------------------------------------------

audit_datasets = {
    "fiqa_train": fiqa_train_normalized,
    "fiqa_validation": fiqa_valid_normalized,
    "fiqa_test": fiqa_test_normalized,

    "twitter_train": twitter_train_normalized,
    "twitter_validation": twitter_valid_normalized,

    "phrasebank": phrasebank_normalized,

    "finance_news": finance_news_normalized
}


# ------------------------------------------------------------
# Normalize text for comparison
# ------------------------------------------------------------

for name, df in audit_datasets.items():

    df["_duplicate_key"] = (
        df["text"]
        .apply(normalize_for_duplicate_check)
    )


# ------------------------------------------------------------
# Function to find overlap
# ------------------------------------------------------------

def find_overlap(df1, df2):

    keys1 = set(df1["_duplicate_key"])
    keys2 = set(df2["_duplicate_key"])

    return keys1.intersection(keys2)


# ------------------------------------------------------------
# Important split comparisons
# ------------------------------------------------------------

comparisons = [
    ("fiqa_train", "fiqa_validation"),
    ("fiqa_train", "fiqa_test"),
    ("fiqa_validation", "fiqa_test"),

    ("twitter_train", "twitter_validation")
]


print("\nSplit leakage checks:\n")


for name1, name2 in comparisons:

    overlap = find_overlap(
        audit_datasets[name1],
        audit_datasets[name2]
    )

    print(
        f"{name1:20} ↔ {name2:20} : "
        f"{len(overlap)} duplicates"
    )
```

```python
# ============================================================
# CELL 31 — CROSS-DATASET OVERLAP
# ============================================================

print("=" * 60)
print("CROSS-DATASET OVERLAP")
print("=" * 60)


dataset_names = list(
    audit_datasets.keys()
)


cross_source_results = []


for i in range(
    len(dataset_names)
):

    for j in range(
        i + 1,
        len(dataset_names)
    ):

        name1 = dataset_names[i]
        name2 = dataset_names[j]

        overlap = find_overlap(
            audit_datasets[name1],
            audit_datasets[name2]
        )

        cross_source_results.append({

            "dataset_1": name1,

            "dataset_2": name2,

            "overlap_count": len(overlap)
        })


cross_source_df = pd.DataFrame(
    cross_source_results
)


display(
    cross_source_df.sort_values(
        "overlap_count",
        ascending=False
    )
)
```

```python
# ============================================================
# CELL 32 — REMOVE INTERNAL DUPLICATES
# ============================================================

print("=" * 60)
print("INTERNAL DUPLICATE REMOVAL")
print("=" * 60)


def remove_internal_duplicates(
    df,
    dataset_name
):

    before = len(df)

    result = (
        df
        .drop_duplicates(
            subset=["text"]
        )
        .reset_index(drop=True)
    )

    after = len(result)

    print(
        f"{dataset_name:20} "
        f"{before:8} → {after:8} "
        f"removed: {before - after}"
    )

    return result


fiqa_train_normalized = remove_internal_duplicates(
    fiqa_train_normalized,
    "FiQA Train"
)

fiqa_valid_normalized = remove_internal_duplicates(
    fiqa_valid_normalized,
    "FiQA Validation"
)

fiqa_test_normalized = remove_internal_duplicates(
    fiqa_test_normalized,
    "FiQA Test"
)

twitter_train_normalized = remove_internal_duplicates(
    twitter_train_normalized,
    "Twitter Train"
)

twitter_valid_normalized = remove_internal_duplicates(
    twitter_valid_normalized,
    "Twitter Validation"
)

phrasebank_normalized = remove_internal_duplicates(
    phrasebank_normalized,
    "PhraseBank"
)

finance_news_normalized = remove_internal_duplicates(
    finance_news_normalized,
    "Finance News"
)
```

```python
# ============================================================
# CELL 33 — CREATE CONTROLLED SPLITS
# ============================================================

from sklearn.model_selection import train_test_split


RANDOM_SEED = 42


def create_stratified_splits(
    df,
    dataset_name
):

    # --------------------------------------------------------
    # First: 80% train / 20% temporary
    # --------------------------------------------------------

    train_df, temp_df = train_test_split(
        df,
        test_size=0.20,
        random_state=RANDOM_SEED,
        stratify=df["label"]
    )


    # --------------------------------------------------------
    # Second: temporary → 10% validation / 10% test
    # --------------------------------------------------------

    validation_df, test_df = train_test_split(
        temp_df,
        test_size=0.50,
        random_state=RANDOM_SEED,
        stratify=temp_df["label"]
    )


    # Reset indices
    train_df = train_df.reset_index(drop=True)
    validation_df = validation_df.reset_index(drop=True)
    test_df = test_df.reset_index(drop=True)


    print(
        f"\n{dataset_name}"
    )

    print(
        f"Train      : {len(train_df)}"
    )

    print(
        f"Validation : {len(validation_df)}"
    )

    print(
        f"Test       : {len(test_df)}"
    )


    return (
        train_df,
        validation_df,
        test_df
    )


# ------------------------------------------------------------
# PhraseBank
# ------------------------------------------------------------

phrasebank_train, \
phrasebank_validation, \
phrasebank_test = create_stratified_splits(
    phrasebank_normalized,
    "FinancialPhraseBank"
)


# ------------------------------------------------------------
# Finance News
# ------------------------------------------------------------

finance_news_train, \
finance_news_validation, \
finance_news_test = create_stratified_splits(
    finance_news_normalized,
    "Finance News"
)
```

```python
# ============================================================
# CELL 34 — FINAL SPLIT DISTRIBUTION
# ============================================================

print("=" * 60)
print("FINAL CLASS DISTRIBUTIONS")
print("=" * 60)


final_splits = {

    "FiQA Train":
        fiqa_train_normalized,

    "FiQA Validation":
        fiqa_valid_normalized,

    "FiQA Test":
        fiqa_test_normalized,

    "Twitter Train":
        twitter_train_normalized,

    "Twitter Validation":
        twitter_valid_normalized,

    "PhraseBank Train":
        phrasebank_train,

    "PhraseBank Validation":
        phrasebank_validation,

    "PhraseBank Test":
        phrasebank_test,

    "Finance News Train":
        finance_news_train,

    "Finance News Validation":
        finance_news_validation,

    "Finance News Test":
        finance_news_test
}


distribution_rows = []


for name, df in final_splits.items():

    distribution_rows.append({

        "split": name,

        "rows": len(df),

        "negative": (
            df["label"] == 0
        ).sum(),

        "neutral": (
            df["label"] == 1
        ).sum(),

        "positive": (
            df["label"] == 2
        ).sum()
    })


final_distribution = pd.DataFrame(
    distribution_rows
)


display(
    final_distribution
)
```

```python
# ============================================================
# CELL 36 — FINAL INTEGRITY CHECK
# ============================================================

print("=" * 60)
print("FINAL DATA INTEGRITY CHECK")
print("=" * 60)


errors = []


for filename, df in final_files.items():

    # Required columns
    expected_columns = [
        "text",
        "label",
        "source"
    ]

    if list(df.columns) != expected_columns:

        errors.append(
            f"{filename}: incorrect columns"
        )


    # Missing text
    if df["text"].isna().any():

        errors.append(
            f"{filename}: missing text"
        )


    # Missing labels
    if df["label"].isna().any():

        errors.append(
            f"{filename}: missing labels"
        )


    # Valid labels
    invalid_labels = (
        ~df["label"].isin([0, 1, 2])
    ).sum()

    if invalid_labels > 0:

        errors.append(
            f"{filename}: "
            f"{invalid_labels} invalid labels"
        )


    # Empty text
    empty_text = (
        df["text"]
        .astype(str)
        .str.strip()
        .eq("")
        .sum()
    )

    if empty_text > 0:

        errors.append(
            f"{filename}: "
            f"{empty_text} empty texts"
        )


# ------------------------------------------------------------
# Report
# ------------------------------------------------------------

if errors:

    print("\n❌ DATA PREPARATION HAS ERRORS:\n")

    for error in errors:
        print(" -", error)

    raise ValueError(
        "Fix data integrity issues before training."
    )


else:

    print(
        "\n✓ No integrity errors found."
    )

    print(
        "✓ All datasets use:"
    )

    print(
        "  text | label | source"
    )

    print(
        "✓ Labels are 0 / 1 / 2"
    )

    print(
        "✓ No missing text"
    )

    print(
        "✓ No missing labels"
    )

    print(
        "✓ Final splits saved"
    )

    print(
        "\nDATA PREPARATION = COMPLETE"
    )
```
