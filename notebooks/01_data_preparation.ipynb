```python
# ============================================================
# CELL 13 — LABEL DISTRIBUTIONS
# ============================================================

def label_distribution(df, name, label_column):

    print("=" * 60)
    print(name)
    print("=" * 60)

    counts = df[label_column].value_counts(
        dropna=False
    )

    percentages = (
        df[label_column]
        .value_counts(
            normalize=True,
            dropna=False
        )
        .mul(100)
        .round(2)
    )

    report = pd.DataFrame({
        "count": counts,
        "percentage": percentages
    })

    display(report)
```

```python
# ============================================================
# CELL 14 — NORMALIZE FiQA
# ============================================================

print("=" * 60)
print("NORMALIZING FiQA")
print("=" * 60)

# ------------------------------------------------------------
# FiQA schema
# ------------------------------------------------------------
# _id
# sentence
# target
# aspect
# score
# type
#
# We use:
# sentence -> text
# score    -> sentiment
# ------------------------------------------------------------


# Create normalized dataframe
fiqa_normalized = fiqa_train[
    ["sentence", "score"]
].copy()


# Rename columns
fiqa_normalized.columns = [
    "text",
    "score"
]


# Add source
fiqa_normalized["source"] = "fiqa"


# ------------------------------------------------------------
# Clean text
# ------------------------------------------------------------

fiqa_normalized["text"] = (
    fiqa_normalized["text"]
    .astype(str)
    .str.strip()
)


# Remove empty text
fiqa_normalized = fiqa_normalized[
    (fiqa_normalized["text"] != "") &
    (fiqa_normalized["text"] != "nan")
].copy()


# ------------------------------------------------------------
# Convert FiQA continuous score
# to common 3-class sentiment
#
# score < -0.1       -> negative = 0
# -0.1 <= score <= .1 -> neutral = 1
# score > 0.1        -> positive = 2
# ------------------------------------------------------------

def fiqa_score_to_label(score):

    if score < -0.1:
        return 0

    elif score <= 0.1:
        return 1

    else:
        return 2


fiqa_normalized["label"] = (
    fiqa_normalized["score"]
    .apply(fiqa_score_to_label)
)


# ------------------------------------------------------------
# Remove original score from final unified schema
# ------------------------------------------------------------

fiqa_normalized = fiqa_normalized[
    ["text", "label", "source"]
].reset_index(drop=True)


# ------------------------------------------------------------
# Verification
# ------------------------------------------------------------

print("\nShape:")
print(fiqa_normalized.shape)

print("\nColumns:")
print(fiqa_normalized.columns.tolist())

print("\nLabel distribution:")
print(
    fiqa_normalized["label"]
    .value_counts()
    .sort_index()
)

print("\nSource distribution:")
print(
    fiqa_normalized["source"]
    .value_counts()
)

print("\nSample:")
display(
    fiqa_normalized.head(10)
)
```

```python
# ============================================================
# CELL 15 — NORMALIZE ALL FiQA SPLITS
# ============================================================

def normalize_fiqa(df, split_name):

    result = df[
        ["sentence", "score"]
    ].copy()

    result.columns = [
        "text",
        "score"
    ]

    result["source"] = "fiqa"

    # Clean text
    result["text"] = (
        result["text"]
        .astype(str)
        .str.strip()
    )

    # Remove empty text
    result = result[
        (result["text"] != "") &
        (result["text"] != "nan")
    ].copy()

    # Convert continuous FiQA score → 3 classes
    result["label"] = result["score"].apply(
        fiqa_score_to_label
    )

    # Final schema
    result = result[
        ["text", "label", "source"]
    ].reset_index(drop=True)

    print(
        f"{split_name:10} → {result.shape}"
    )

    return result


# Normalize each split separately

fiqa_train_normalized = normalize_fiqa(
    fiqa_train,
    "Train"
)

fiqa_valid_normalized = normalize_fiqa(
    fiqa_valid,
    "Validation"
)

fiqa_test_normalized = normalize_fiqa(
    fiqa_test,
    "Test"
)
```

```python
# ============================================================
# CELL 18 — NORMALIZE TWITTER
# ============================================================

TWITTER_TEXT_COLUMN = "text"
TWITTER_LABEL_COLUMN = "label"


def normalize_twitter(df, split_name):

    result = df[
        [
            TWITTER_TEXT_COLUMN,
            TWITTER_LABEL_COLUMN
        ]
    ].copy()

    result.columns = [
        "text",
        "label"
    ]

    result["source"] = "twitter"

    # Clean text
    result["text"] = (
        result["text"]
        .astype(str)
        .str.strip()
    )

    # Remove empty text
    result = result[
        (result["text"] != "") &
        (result["text"] != "nan")
    ].copy()

    # --------------------------------------------------------
    # Handle labels
    # --------------------------------------------------------

    # If labels are already numeric 0/1/2
    if pd.api.types.is_numeric_dtype(
        result["label"]
    ):

        result["label"] = (
            result["label"]
            .astype(int)
        )

    # If labels are strings
    else:

        result["label"] = (
            result["label"]
            .astype(str)
            .str.lower()
            .str.strip()
            .map({
                "negative": 0,
                "neutral": 1,
                "positive": 2
            })
        )

    # Check for invalid labels
    invalid = result[
        ~result["label"].isin([0, 1, 2])
    ]

    if len(invalid) > 0:

        print(
            f"⚠ {split_name}: "
            f"{len(invalid)} invalid labels found"
        )

        print(
            invalid["label"]
            .value_counts(dropna=False)
        )

        raise ValueError(
            "Twitter label mapping needs to be checked."
        )

    result["label"] = (
        result["label"]
        .astype(int)
    )

    return result[
        ["text", "label", "source"]
    ].reset_index(drop=True)


twitter_train_normalized = normalize_twitter(
    twitter_train,
    "Train"
)

twitter_valid_normalized = normalize_twitter(
    twitter_valid,
    "Validation"
)


print(
    "Twitter Train:",
    twitter_train_normalized.shape
)

print(
    "Twitter Validation:",
    twitter_valid_normalized.shape
)
```

```python
# ============================================================
# CELL 21 — NORMALIZE FINANCIALPHRASEBANK
# ============================================================

phrasebank_records = []

for line in phrasebank_lines:

    line = line.strip()

    if not line:
        continue

    # Sentence and sentiment are separated by @
    parts = line.rsplit("@", 1)

    if len(parts) != 2:
        continue

    text, sentiment = parts

    text = text.strip()
    sentiment = sentiment.strip().lower()

    phrasebank_records.append({
        "text": text,
        "sentiment": sentiment
    })


phrasebank_normalized = pd.DataFrame(
    phrasebank_records
)


print("Parsed rows:", phrasebank_normalized.shape)

print("\nOriginal labels:")
print(
    phrasebank_normalized["sentiment"]
    .value_counts()
)


# ------------------------------------------------------------
# Map sentiment → common HACE labels
# ------------------------------------------------------------

phrasebank_normalized["label"] = (
    phrasebank_normalized["sentiment"]
    .map({
        "negative": 0,
        "neutral": 1,
        "positive": 2
    })
)


# Check invalid labels
invalid = phrasebank_normalized[
    phrasebank_normalized["label"].isna()
]

print("\nInvalid labels:", len(invalid))

if len(invalid) > 0:
    print(invalid["sentiment"].value_counts())


# Remove sentiment text column
phrasebank_normalized = (
    phrasebank_normalized[
        ["text", "label"]
    ]
    .copy()
)

phrasebank_normalized["source"] = "phrasebank"

phrasebank_normalized["label"] = (
    phrasebank_normalized["label"]
    .astype(int)
)

phrasebank_normalized = (
    phrasebank_normalized
    .reset_index(drop=True)
)


print("\nFinal shape:")
print(phrasebank_normalized.shape)

display(
    phrasebank_normalized.head()
)
```

```python
# ============================================================
# CELL 24 — NORMALIZE KAGGLE FINANCE NEWS
# ============================================================

print("=" * 60)
print("NORMALIZING FINANCE NEWS")
print("=" * 60)


# ------------------------------------------------------------
# Create normalized dataframe
# ------------------------------------------------------------

finance_news_normalized = finance_news[
    ["text", "sentiment"]
].copy()


# Rename sentiment → label
finance_news_normalized = (
    finance_news_normalized
    .rename(
        columns={
            "sentiment": "label"
        }
    )
)


# Add source
finance_news_normalized["source"] = (
    "finance_news"
)


# ------------------------------------------------------------
# Clean text
# ------------------------------------------------------------

# Convert to string
finance_news_normalized["text"] = (
    finance_news_normalized["text"]
    .astype("string")
    .str.strip()
)


# Remove missing / empty text
finance_news_normalized = (
    finance_news_normalized[
        finance_news_normalized["text"].notna()
    ]
    .copy()
)

finance_news_normalized = (
    finance_news_normalized[
        finance_news_normalized["text"] != ""
    ]
    .copy()
)


# ------------------------------------------------------------
# Convert labels
# ------------------------------------------------------------

finance_news_normalized["label"] = (
    finance_news_normalized["label"]
    .astype(str)
    .str.lower()
    .str.strip()
    .map({
        "negative": 0,
        "neutral": 1,
        "positive": 2
    })
)


# ------------------------------------------------------------
# Verify label conversion
# ------------------------------------------------------------

print("\nLabel distribution:")

print(
    finance_news_normalized["label"]
    .value_counts(dropna=False)
    .sort_index()
)


# Check invalid labels
invalid_labels = (
    finance_news_normalized["label"]
    .isna()
    .sum()
)

print(
    "\nInvalid / unmapped labels:",
    invalid_labels
)


if invalid_labels > 0:

    raise ValueError(
        "Some Finance News labels could not "
        "be mapped to 0/1/2."
    )


finance_news_normalized["label"] = (
    finance_news_normalized["label"]
    .astype(int)
)


# ------------------------------------------------------------
# Final schema
# ------------------------------------------------------------

finance_news_normalized = (
    finance_news_normalized[
        ["text", "label", "source"]
    ]
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# Final verification
# ------------------------------------------------------------

print("\nFinal shape:")
print(finance_news_normalized.shape)

print("\nColumns:")
print(finance_news_normalized.columns.tolist())

print("\nMissing values:")
print(finance_news_normalized.isnull().sum())

print("\nSamples:")
display(
    finance_news_normalized.head(10)
)
```
