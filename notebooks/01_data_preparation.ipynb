```python
# ============================================================
# HACE — DATA PREPARATION
# CELL 1: IMPORTS
# ============================================================

import sys
from pathlib import Path

import pandas as pd
import numpy as np

print("Python:", sys.version)
print("Pandas:", pd.__version__)
print("NumPy:", np.__version__)

print("\n✓ Imports successful")
```

```python
# ============================================================
# CELL 2: PROJECT PATHS
# ============================================================

PROJECT_ROOT = Path(
    "/home/amankhanduja/NLP_Project/HACE"
)

RAW_DIR = PROJECT_ROOT / "data" / "raw"
INTERIM_DIR = PROJECT_ROOT / "data" / "interim"
PROCESSED_DIR = PROJECT_ROOT / "data" / "processed"

FIQA_DIR = RAW_DIR / "fiqa"
TWITTER_DIR = RAW_DIR / "twitter"
PHRASEBANK_DIR = RAW_DIR / "phrasebank"
FINANCE_NEWS_DIR = RAW_DIR / "finance_news"


# Create output directories
INTERIM_DIR.mkdir(
    parents=True,
    exist_ok=True
)

PROCESSED_DIR.mkdir(
    parents=True,
    exist_ok=True
)


print("Project Root :", PROJECT_ROOT)
print("Raw Data     :", RAW_DIR)
print("Interim      :", INTERIM_DIR)
print("Processed    :", PROCESSED_DIR)
```

```python
# ============================================================
# CELL 3: VERIFY RAW DATA
# ============================================================

datasets = {
    "FiQA": FIQA_DIR,
    "Twitter": TWITTER_DIR,
    "FinancialPhraseBank": PHRASEBANK_DIR,
    "Finance News": FINANCE_NEWS_DIR
}


for name, directory in datasets.items():

    print("\n" + "=" * 50)
    print(name)
    print("=" * 50)

    if not directory.exists():

        print("❌ Directory does not exist:")
        print(directory)
        continue

    files = list(directory.iterdir())

    if not files:

        print("⚠ Directory is empty")
        continue

    for file in files:

        if file.is_file():

            size_mb = (
                file.stat().st_size /
                (1024 * 1024)
            )

            print(
                f"✓ {file.name:35} "
                f"{size_mb:.2f} MB"
            )
```

```python
# ============================================================
# CELL 5: LOAD TWITTER FINANCIAL NEWS
# ============================================================

twitter_train = pd.read_csv(
    TWITTER_DIR / "train.csv"
)

twitter_valid = pd.read_csv(
    TWITTER_DIR / "validation.csv"
)


print("Twitter Train:", twitter_train.shape)
print("Twitter Valid:", twitter_valid.shape)

print("\nColumns:")
print(twitter_train.columns.tolist())

print("\nSample:")
display(twitter_train.head())
```

```python
# ============================================================
# CELL 6: LOAD FINANCIALPHRASEBANK
# ============================================================

phrasebank_file = (
    PHRASEBANK_DIR /
    "Sentences_AllAgree.txt"
)


print("File:", phrasebank_file)


with open(
    phrasebank_file,
    "r",
    encoding="latin-1"
) as f:

    phrasebank_lines = f.readlines()


print(
    "Number of lines:",
    len(phrasebank_lines)
)


print("\nFirst 5 lines:")

for line in phrasebank_lines[:5]:

    print(line.strip())
```

```python
# ============================================================
# CELL 8: INSPECT LABELS
# ============================================================

print("=" * 60)
print("FiQA LABELS")
print("=" * 60)

print(fiqa_train.columns.tolist())

for column in fiqa_train.columns:

    print(
        f"\n{column}:"
    )

    print(
        fiqa_train[column]
        .value_counts(dropna=False)
        .head(20)
    )


print("\n" + "=" * 60)
print("TWITTER LABELS")
print("=" * 60)

for column in twitter_train.columns:

    print(
        f"\n{column}:"
    )

    print(
        twitter_train[column]
        .value_counts(dropna=False)
        .head(20)
    )


print("\n" + "=" * 60)
print("FINANCE NEWS LABELS")
print("=" * 60)

for column in finance_news.columns:

    print(
        f"\n{column}:"
    )

    print(
        finance_news[column]
        .value_counts(dropna=False)
        .head(20)
    )
```

```python
# ============================================================
# CELL 9 — MISSING VALUE ANALYSIS
# ============================================================

def missing_report(df, name):
    print("=" * 60)
    print(name)
    print("=" * 60)

    report = pd.DataFrame({
        "missing_count": df.isnull().sum(),
        "missing_percent": (
            df.isnull().mean() * 100
        ).round(2)
    })

    display(report)


missing_report(fiqa_train, "FiQA Train")
missing_report(twitter_train, "Twitter Train")
missing_report(finance_news, "Finance News")

print("=" * 60)
print("FINANCIALPHRASEBANK")
print("=" * 60)

print("Total lines:", len(phrasebank_lines))

empty_lines = [
    line for line in phrasebank_lines
    if not line.strip()
]

print("Empty lines:", len(empty_lines))
```

```python
# ============================================================
# CELL 10 — DUPLICATE ANALYSIS
# ============================================================

def duplicate_report(df, name, text_column):
    
    print("=" * 60)
    print(name)
    print("=" * 60)

    total = len(df)

    duplicates = df[text_column].duplicated().sum()

    unique = df[text_column].nunique()

    print("Total rows      :", total)
    print("Unique texts    :", unique)
    print("Duplicate rows  :", duplicates)
    print(
        "Duplicate %     :",
        round((duplicates / total) * 100, 2),
        "%"
    )


# IMPORTANT:
# Use the actual text-column names you discovered in Cell 8.
```

```python
duplicate_report(
    fiqa_train,
    "FiQA Train",
    "sentence"
)

duplicate_report(
    twitter_train,
    "Twitter Train",
    "text"
)

duplicate_report(
    finance_news,
    "Finance News",
    "text"
)
```

```python
# ============================================================
# CELL 11 — TEXT STATISTICS
# ============================================================

def text_statistics(df, name, text_column):

    text_lengths = (
        df[text_column]
        .astype(str)
        .str.split()
        .str.len()
    )

    print("=" * 60)
    print(name)
    print("=" * 60)

    print("Number of samples :", len(df))
    print("Average words     :", round(text_lengths.mean(), 2))
    print("Median words      :", round(text_lengths.median(), 2))
    print("Maximum words     :", text_lengths.max())
    print("Minimum words     :", text_lengths.min())


# Example:
# text_statistics(fiqa_train, "FiQA", "sentence")
# text_statistics(twitter_train, "Twitter", "text")
# text_statistics(finance_news, "Finance News", "text")
```

```python
# ============================================================
# CELL 13 — LABEL DISTRIBUTIONS
# ============================================================

def label_distribution(df, name, label_column):

    print("=" * 60)
    print(name)
    print("=" * 60)

    counts = df[label_column].value_counts(
        dropna=False
    )

    percentages = (
        df[label_column]
        .value_counts(
            normalize=True,
            dropna=False
        )
        .mul(100)
        .round(2)
    )

    report = pd.DataFrame({
        "count": counts,
        "percentage": percentages
    })

    display(report)
```
