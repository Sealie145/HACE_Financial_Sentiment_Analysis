```python
import os
import glob
import shutil
import zipfile
import subprocess
from pathlib import Path

import pandas as pd
from datasets import load_dataset
```

```python
# ============================================================
# CELL 2 — PROJECT PATHS
# ============================================================

PROJECT_ROOT = Path("/home/amankhanduja/NLP_Project/HACE")

RAW_DIR = PROJECT_ROOT / "data" / "raw"

FIQA_DIR = RAW_DIR / "fiqa"
PHRASEBANK_DIR = RAW_DIR / "phrasebank"
TWITTER_DIR = RAW_DIR / "twitter"
FINANCE_NEWS_DIR = RAW_DIR / "finance_news"


# Create directories
for directory in [
    RAW_DIR,
    FIQA_DIR,
    PHRASEBANK_DIR,
    TWITTER_DIR,
    FINANCE_NEWS_DIR
]:
    directory.mkdir(
        parents=True,
        exist_ok=True
    )


print("=" * 60)
print("HACE PROJECT PATHS")
print("=" * 60)

print(f"Project Root : {PROJECT_ROOT}")
print(f"Raw Data     : {RAW_DIR}")

print("\nDirectories ready:")
print(f"FiQA         : {FIQA_DIR}")
print(f"PhraseBank   : {PHRASEBANK_DIR}")
print(f"Twitter      : {TWITTER_DIR}")
print(f"Finance News : {FINANCE_NEWS_DIR}")
```

```python
# ============================================================
# CELL 3 — DOWNLOAD FiQA
# ============================================================

print("=" * 60)
print("DOWNLOADING FiQA")
print("=" * 60)

fiqa = load_dataset(
    "TheFinAI/fiqa-sentiment-classification"
)

print("\nAvailable splits:")
print(fiqa)


for split in fiqa.keys():

    df = fiqa[split].to_pandas()

    output_path = FIQA_DIR / f"{split}.csv"

    df.to_csv(
        output_path,
        index=False
    )

    print(
        f"✓ {split:12} "
        f"{str(df.shape):15} → {output_path}"
    )


print("\n✓ FiQA download complete")
```

```python
# ============================================================
# CELL 4 — DOWNLOAD TWITTER FINANCIAL NEWS
# ============================================================

print("=" * 60)
print("DOWNLOADING TWITTER FINANCIAL NEWS")
print("=" * 60)

twitter = load_dataset(
    "zeroshot/twitter-financial-news-sentiment"
)

print("\nAvailable splits:")
print(twitter)


for split in twitter.keys():

    df = twitter[split].to_pandas()

    output_path = TWITTER_DIR / f"{split}.csv"

    df.to_csv(
        output_path,
        index=False
    )

    print(
        f"✓ {split:12} "
        f"{str(df.shape):15} → {output_path}"
    )


print("\n✓ Twitter dataset download complete")
```

```python
# ============================================================
# CELL 5 — FINANCIALPHRASEBANK
# ============================================================

print("=" * 60)
print("SETTING UP FINANCIALPHRASEBANK")
print("=" * 60)


phrasebank_destination = (
    PHRASEBANK_DIR / "Sentences_AllAgree.txt"
)


# Possible locations
possible_sources = [

    # Notebook directory
    Path.cwd() / "Sentences_AllAgree.txt",

    # HACE root
    PROJECT_ROOT / "Sentences_AllAgree.txt",

    # Already in raw/phrasebank
    phrasebank_destination,

    # Downloads
    Path.home()
    / "Downloads"
    / "FinancialPhraseBank-v1.0"
    / "FinancialPhraseBank-v1.0"
    / "Sentences_AllAgree.txt",

    Path.home()
    / "Downloads"
    / "FinancialPhraseBank-v1.0"
    / "Sentences_AllAgree.txt"
]


# Check whether destination already exists
if phrasebank_destination.exists():

    print(
        f"✓ Already exists:\n"
        f"  {phrasebank_destination}"
    )


else:

    source_found = None

    for source in possible_sources:

        if source.exists():

            source_found = source
            break


    if source_found is not None:

        shutil.copy2(
            source_found,
            phrasebank_destination
        )

        print(
            "✓ FinancialPhraseBank copied"
        )

        print(
            f"From: {source_found}"
        )

        print(
            f"To:   {phrasebank_destination}"
        )


    else:

        print(
            "⚠ Sentences_AllAgree.txt not found."
        )

        print(
            "\nPlease place the file here:"
        )

        print(
            phrasebank_destination
        )
```

```python
# ============================================================
# CELL 6 — DOWNLOAD KAGGLE FINANCE NEWS SENTIMENTS
# ============================================================

print("=" * 60)
print("DOWNLOADING KAGGLE FINANCE NEWS SENTIMENTS")
print("=" * 60)


# ------------------------------------------------------------
# Check whether CSV already exists
# ------------------------------------------------------------

existing_csvs = list(
    FINANCE_NEWS_DIR.glob("*.csv")
)


if existing_csvs:

    print("✓ Finance News dataset already exists:")

    for csv_file in existing_csvs:

        print(
            f"  {csv_file.name}"
        )


else:

    print(
        "No Finance News CSV found."
    )

    print(
        "Attempting Kaggle download..."
    )


    # --------------------------------------------------------
    # Find Kaggle CLI
    # --------------------------------------------------------

    kaggle_command = shutil.which("kaggle")


    if kaggle_command is None:

        print(
            "Kaggle CLI not found."
        )

        print(
            "Installing Kaggle..."
        )

        subprocess.run(
            [
                "python",
                "-m",
                "pip",
                "install",
                "-q",
                "kaggle"
            ],
            check=True
        )

        kaggle_command = shutil.which("kaggle")


    # --------------------------------------------------------
    # Download Kaggle dataset
    # --------------------------------------------------------

    result = subprocess.run(
        [
            kaggle_command,
            "datasets",
            "download",
            "-d",
            "antobenedetti/finance-news-sentiments",
            "-p",
            str(FINANCE_NEWS_DIR)
        ],
        capture_output=True,
        text=True
    )


    # --------------------------------------------------------
    # Check result
    # --------------------------------------------------------

    if result.returncode != 0:

        print("\n⚠ Kaggle download failed.")

        print("\nSTDOUT:")
        print(result.stdout)

        print("\nSTDERR:")
        print(result.stderr)

        print(
            "\nCheck your Kaggle authentication."
        )


    else:

        print(
            "✓ Kaggle dataset downloaded"
        )


        # ----------------------------------------------------
        # Extract ZIP
        # ----------------------------------------------------

        zip_files = list(
            FINANCE_NEWS_DIR.glob("*.zip")
        )


        if not zip_files:

            print(
                "⚠ No ZIP file found."
            )


        else:

            for zip_path in zip_files:

                print(
                    f"Extracting: {zip_path.name}"
                )

                with zipfile.ZipFile(
                    zip_path,
                    "r"
                ) as zip_ref:

                    zip_ref.extractall(
                        FINANCE_NEWS_DIR
                    )


                # Delete ZIP after extraction
                zip_path.unlink()

                print(
                    "✓ Extracted"
                )


print("\n✓ Finance News setup complete")
```

```python
# ============================================================
# CELL 7 — VERIFY RAW DATA
# ============================================================

print("=" * 60)
print("HACE RAW DATA VERIFICATION")
print("=" * 60)


dataset_dirs = [
    FIQA_DIR,
    PHRASEBANK_DIR,
    TWITTER_DIR,
    FINANCE_NEWS_DIR
]


for dataset_dir in dataset_dirs:

    print(
        f"\n[{dataset_dir.name.upper()}]"
    )

    files = list(
        dataset_dir.iterdir()
    )


    if not files:

        print("  ⚠ EMPTY")

        continue


    for file in files:

        if file.is_file():

            size_mb = (
                file.stat().st_size
                / (1024 * 1024)
            )

            print(
                f"  ✓ {file.name:35}"
                f"{size_mb:.2f} MB"
            )


# ============================================================
# Final file listing
# ============================================================

print("\n" + "=" * 60)
print("FINAL RAW DATA STRUCTURE")
print("=" * 60)


for path in sorted(RAW_DIR.rglob("*")):

    if path.is_file():

        print(
            path.relative_to(PROJECT_ROOT)
        )


print("\n" + "=" * 60)
print("✓ RAW DATA SETUP COMPLETE")
print("=" * 60)
```
